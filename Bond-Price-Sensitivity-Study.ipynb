```python
def calculate_bond_price(face_value, coupon_rate, yield_rate, maturity):
    """
    When entering a rate input, the rate must be in decimals for the purposes of this model :
    10% = 0.1
    """
    annual_coupon = face_value * coupon_rate

    coupon_values = sum(
        annual_coupon / (1 + yield_rate)**i
        for i in range(1, maturity + 1)
    )

    redemption_value = face_value / (1 + yield_rate)**maturity

    return coupon_values + redemption_value

#I have given an example of a premium coupon bond; This is a bond where the market price is higher than its face value.
# Premium coupon bond
price = calculate_bond_price(
    face_value=1000,
    coupon_rate=0.07,
    yield_rate=0.045,
    maturity=10
)

print(f"Bond price: £{price:.2f}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

yields = np.arange(0.01, 0.101, 0.01)

results = []

'''
The question we are trying to answer here is how changes in market interest rates (yields) affect the price of a bond.

This time we are using a simple 5-year fixed-rate coupon bond.

'''
for yield_rate in yields:
    price = calculate_bond_price(
        face_value=1000,
        coupon_rate=0.05,
        yield_rate=yield_rate,
        maturity=5
    )

    results.append({
        "Yield (%)": yield_rate * 100,
        "Bond Price (£)": price
    })

df = pd.DataFrame(results)

print(df.round(2).to_string(index=False))

# Therefore, we can see now how increasing yields leads to decreasing principal value of the Bond 
# This clearly demonstrates the inverse relationship between bond prices & yields
```

```python
import pandas as pd
import numpy as n
import matplotlib.pyplot as plt

maturities = [2, 5, 10, 20]

#Here we are giving examples of Short-Term , Intermediate-Term & Long-Term Bonds

yields = n.arange(0.01, 0.101, 0.01)

plt.figure(figsize=(9, 6))

for maturity in maturities:
    prices = [
        calculate_bond_price(
            face_value=1000,
            coupon_rate=0.025,
            yield_rate=y,
            maturity=maturity
        )
        for y in yields
    ]

    plt.plot(
        yields * 100,
        prices,
        marker="o",
        label=f"{maturity}-year bond"
    )
    
plt.ylabel("Bond Price (Principle) (£)")
plt.xlabel("Yield (%)")
plt.title("Interest Rate Sensitivity to Maturity")
plt.legend()
plt.grid(True)
plt.show()


#This is also showing a point that all the bonds of different maturities converge to which is 4% which was the input simple-coupon bond earlier

#When the market yield is also 4%, the price of every bond equals its face value (£1,000).

#This happens regardless of whether the bond matures in 2, 5, 10 or 20 years.

#Which is why they converge to that common point

```
